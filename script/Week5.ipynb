## Part 1: Load Final Datasets

Load the cleaned datasets generated from Week 4.

These datasets already include:

- Feature selection
- Missing value handling
- Date conversion
- Date consistency flags

```python
import pandas as pd
import numpy as np

listings = pd.read_csv(
    "../outputs/final_clean_listings.csv",
    low_memory=False
)

sold = pd.read_csv(
    "../outputs/final_clean_sold.csv",
    low_memory=False
)

print("Listings:", listings.shape)
print("Sold:", sold.shape)
```

```python
for df in [listings, sold]:

    df["missing_coordinate_flag"] = (
        df["Latitude"].isna() |
        df["Longitude"].isna()
    )

print("Listings Missing Coordinates:",
      listings["missing_coordinate_flag"].sum())

print("Sold Missing Coordinates:",
      sold["missing_coordinate_flag"].sum())
```

```python
for df in [listings, sold]:

    df["zero_coordinate_flag"] = (
        (df["Latitude"] == 0) |
        (df["Longitude"] == 0)
    )

print("Listings Zero Coordinates:",
      listings["zero_coordinate_flag"].sum())

print("Sold Zero Coordinates:",
      sold["zero_coordinate_flag"].sum())
```

```python
for df in [listings, sold]:

    df["positive_longitude_flag"] = (
        df["Longitude"] > 0
    )

print("Listings Positive Longitude:",
      listings["positive_longitude_flag"].sum())

print("Sold Positive Longitude:",
      sold["positive_longitude_flag"].sum())
```

```python
for df in [listings, sold]:

    df["implausible_coordinate_flag"] = (
        (df["Latitude"] < 32) |
        (df["Latitude"] > 42) |
        (df["Longitude"] < -125) |
        (df["Longitude"] > -114)
    )


print("Listings Implausible Coordinates:",
      listings["implausible_coordinate_flag"].sum())

print("Sold Implausible Coordinates:",
      sold["implausible_coordinate_flag"].sum())
```
